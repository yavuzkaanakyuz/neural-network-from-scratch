```python
import matplotlib.pyplot as plt

expected = 1.0

x = 1.0

weights = [-2, -1, 0, 0.5, 1, 1.5, 2, 3]

losses = []


def loss_function(predicted, expected):
    return (predicted - expected) ** 2


for w in weights:
    predicted = x * w

    loss = loss_function(predicted, expected)
    losses.append(loss)

    print("Weight:", w, "Prediction:", predicted, "Loss:", loss)


plt.plot(weights, losses, marker="o")

plt.xlabel("Weight")
plt.ylabel("Loss")
plt.title("Loss Curve")

plt.show()
```
