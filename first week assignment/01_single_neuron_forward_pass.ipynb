```python
inputs = [1.00, 0.25, 0.25]

weights = [1.2, 0.5, 0.8]

bias = -1.5

def exp(x):
  result = 1.0
  term = 1.0

  for n in range(1, 30):
    term *= x/n
    result += term
  return result

def sigmoid(x):
  return 1 / (1 + exp(-x))

def neuron_forward(inputs, weights, bias):
  z = 0.0

  for x, w in zip(inputs, weights):
    z += x * w

  z += bias
  output = sigmoid(z)

  return z, output

z, output = neuron_forward(inputs, weights, bias)

print("Weighted sum: ", z)
print("Neuron output: ", output)
```
