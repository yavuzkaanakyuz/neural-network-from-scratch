```python
x = 2.0
expected = 4.0

w = 0.0

h = 0.0001

learning_rate = 0.1


def loss_function(w):
    predicted = x * w
    return (predicted - expected) ** 2


def numerical_derivative(w):
    return (loss_function(w + h) - loss_function(w)) / h


for step in range(10):

    loss = loss_function(w)

    gradient = numerical_derivative(w)

    print(
        "Step:", step,
        "Weight:", round(w, 4),
        "Loss:", round(loss, 4),
        "Gradient:", round(gradient, 4)
    )

    w = w - learning_rate * gradient


print("\nFinal weight:", w)
print("Final loss:", loss_function(w))
```
