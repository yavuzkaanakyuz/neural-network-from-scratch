```python
inputs = [1.00, 0.25, 0.25]

weights = [
    [1.2, 0.5, 0.8],
    [0.7, 1.1, 0.4],
    [0.3, 0.6, 1.5]
]

biases = [-1.5, -1.0, -1.2]

def exp(x):
  result = 1.0
  term = 1.0

  for n in range(1, 30):
    term *= x / n
    result += term

  return result

def sigmoid(x):
  return 1 / (1 + exp(-x))

def layer_forward(inputs, weights, biases):
  outputs = []

  for neuron_weights, bias in zip(weights, biases):
    z = 0.0

    for x, w in zip(inputs, neuron_weights):
      z += x * w

    z += bias

    output = sigmoid(z)
    outputs.append(output)

  return outputs

outputs = layer_forward(inputs, weights, biases)
print("Layer outputs:", outputs)
```
